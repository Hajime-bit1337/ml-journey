```python
import pandas as pd

Data1 = pd.DataFrame({
    'Name': ['Рома', 'Матвей', 'Дима', 'Ливес', 'Ксюша'],
    'Age': [18, 18, 23, 19, 17],
    'Money': [10000000, 10000, 3123, 999, 0],
    'City': ['Санкт-Петербург', 'Москва', 'Казань', 'Гамбург', 'Москва'],
    'IQ': [145, 55, 100, 122, 90]
})
Data1
```

```python
Data2 = pd.DataFrame({
    'Name': ['John', 'Max', 'Eduard', 'Kenny', 'Steve'],
    'Age': [24, 15, 32, 15, 17],
    'Money': [12132, 500, 30000000, 1000, 50],
    'City': ['New-York', 'London', 'Turkey', 'London', 'Paris'],
    'IQ': [120, 121, 92, 122, 100]
})

Data2
```

Концепект по 5:

dtype - показывает тип выбранного столбца пример - reviews.price.dtype
        можно еще использовать сразу для всех столбцов - reviews.dtypes

```python
print(Data1.Money.dtype)
Data2.dtypes
```

astype - используется уже не для просмотра типа а для его задавания - reviews.price.astype('float64')

```python
Data1.Money.astype('float64')
print(Data1.Money.dtype)
```

NaN — это маркер пустоты, который технически всегда является числом с плавающей точкой, то есть имеет тип float64.

```python
for_ex = pd.DataFrame({
    'NaNs': [float('nan'), float('nan'), float('nan')]
})
print(for_ex.NaNs.dtype)

pd.Series([1, 2, float('nan')])
```

isnull / notnull - методы предназначенные для работы с NaN'ами - reviews[pd.isnull(reviews.country)]

```python
for_ex[pd.isnull(for_ex.NaNs)]
```

fillna - метод для replacing missing data - reviews.region_2.fillna("Unknown")
Альтернатива - reviews.taster_twitter_handle.replace("@kerinokeefe", "@kerino")

```python
for_ex.NaNs.fillna('Hello')
```

Концпект по 6:

rename - позволяет менять имена у столбцов или строк
Примеры:
reviews.rename(columns={'points': 'score'})
reviews.rename(index={0: 'firstEntry', 1: 'secondEntry'})

```python
data_coins = Data1.rename(columns={'Money':'coins'})
data_coins
```

set_index - ставит колонку в качестве индекса

```python
data_coins2 = Data1.rename(columns={'Money': 'coins'}).set_index('coins')
data_coins2
```

rename_axis - нужна того чтобы менять атрибуты индексов - reviews.rename_axis("wines", axis='rows').rename_axis("fields", axis='columns')

```python
data_coins3 = data_coins2.rename_axis('money', axis='rows').rename_axis('stats', axis='columns')
data_coins3
```

Для того чтобы комбинировать разные датасеты существуют merge() concat() join(), также написано что большинство функций функции merge() можно упростить с помощью функции join() поэтому сфокусируемся на двух оставшихся способах

concat - сжимает датасеты по рядам, а также можно по столбцам при помощи параметра axis=1 - pd.concat([canadian_youtube, british_youtube])

```python
Data3 = pd.concat([Data1, Data2], axis=1)
Data3
```

join - сжимает датасеты по общим индексам и является следующим по сложности - left.join(right, lsuffix='_CAN', rsuffix='_UK')

```python
countries = pd.DataFrame({
    'City': ['Москва', 'Санкт-Петербург', 'Казань', 'Гамбург'],
    'Country': ['Россия', 'Россия', 'Россия', 'Германия']
}).set_index('City')

Data4 = Data1.set_index('City').join(countries, how='left')

Data4
```

Изначально Матвей и Ксюша дублировались из за того что я прописал Москву в countries дважды, а также у ливеса в был NaN в столбце country из за того что при создании countries я допустил опечатку в названии города Гамбург
