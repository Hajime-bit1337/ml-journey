Концепект по kaggle 5-6:

Концепект по 5:

dtype - показывает тип выбранного столбца пример - reviews.price.dtype
        можно еще использовать сразу для всех столбцов - reviews.dtype

astype - используется уже не для просмотра типа а для его задавания - reviews.price.astype(float64)

NaN - not a number по тех причинам он всегда float64 dtype

isnull / isnotnull - методы предназначенные для работы с NaN'ами - reviews[pd.isnull(reviews.country)]

fillna - метод для replacing missing data - reviews.region_2.fillna("Unknown")
Альтернатива - reviews.taster_twitter_handle.replace("@kerinokeefe", "@kerino")

Концпект по 6:

rename - позволяет менять имена у столбцов или строк
Примеры:
reviews.rename(columns={'points': 'score'})
reviews.rename(index={0: 'firstEntry', 1: 'secondEntry'})

set_index - более удобный способ менять индексы чем renamee

rename_axis - нужна того чтобы менять атрибуты индексов - reviews.rename_axis("wines", axis='rows').rename_axis("fields", axis='columns')

Для того чтобы комбинировать разные датасеты существуют merge() concat() join(), также написано что все что можно сделать merge() можно сделать при помощи join() поэтому сфокусируемся на двух оставшихся способах

concat - сжимает датасеты по столбцам - pd.concat([canadian_youtube, british_youtube])

join - сжимает датасеты по общим индексам и является следующим по сложности - left.join(right, lsuffix='_CAN', rsuffix='_UK')
