Ожидаю 7 строк

У merge есть параметр, который проверяет ожидаемую кратность связи и падает с ошибкой, если ключ оказался неуникальным там, где ты этого не ждал. Найди его и примени к задаче 2, объявив, что связь «многие к одному». Потом попробуй объявить «один к одному» и посмотри, что произойдёт

```python
orders.merge(customers, left_on='customer_id', right_on='id', how='inner', validate='m:1')
```

1 к 1 падает с ошшибкой поэтому я убрал, а m:1 ничего не меняет, просто провалидировал

У merge есть параметр, который добавляет колонку с пометкой, из какой таблицы пришла строка: только слева, только справа или из обеих. Найди его и примени к задаче 3 — увидишь клиента без заказов сразу, без отдельного запроса

```python
customers.merge(orders, left_on='id', right_on='customer_id', how='left', indicator=True)
```

При помощи этого параметра можно заметить единственную строку, которая пришла только из левой таблицы и которая является ответом

Вопрос без кода: чем crosstab отличается от pivot_table?
Ответ: Первое что я заметил crosstab по умолчанию счиатет количество просто или можно задать среднее значение, когда в pivot_table можно передать разынае функции. Второе, crosstab вызываеется как метод pandas и в него передаются Series, а в pivot_table вызывается как метод DataFrame и в него передаются уже его индексы
