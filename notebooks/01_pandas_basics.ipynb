```python
import pandas as pd
pd.__version__
```

Создаю датафрейм:

```python
Data = pd.DataFrame({
    'Name': ['Рома', 'Матвей', 'Дима', 'Ливес', 'Ксюша'],
    'Age': [18, 18, 23, 19, 17],
    'Money': [10000000, 10000, 3123, 999, 0],
    'City': ['Санкт-Петербург', 'Москва', 'Казань', 'Гамбург', 'Москва'],
    'IQ': [145, 55, 100, 122, 90]
})

print(Data)
```

Вывожу информацию об Романе через iloc:

```python
RomanInfo = Data.iloc[0, :4]
print(RomanInfo)
```

Вывожу информацию об Матвее через loc:

```python
MatveyInfo = Data.loc[1, 'Name':'City']
print(MatveyInfo)
```

Вывожу информацию о богатых:

```python
Rich = Data[Data['Money'] > 1000]
print(Rich)
```

Групирую инфу об среднем айкью по возврасту:

```python
Smart = Data.groupby('Age').IQ.mean()
print(Smart)
```

Сортирую информацию о богатых:

```python
RichSort = Rich.sort_values(by = 'Money')
print(RichSort)
```
