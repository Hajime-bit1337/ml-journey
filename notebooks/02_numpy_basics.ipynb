```python
import numpy as np
import time
rng = np.random.default_rng(42)
X = rng.integers(1, 100, size=(1000, 5)).astype('float64')
```

1. **Циклами.** Реализуй нормализацию через `for` по столбцам и строкам, без векторных операций. Замерь время через `time.perf_counter()`.
2. **Векторно.** То же самое одним выражением, без единого `for`. Понадобятся `X.mean(axis=?)` и `X.std(axis=?)` — какой axis, решай по правилу «сколько чисел должно получиться».
3. **Сравни.** Проверь, что результаты совпадают: `np.allclose(a, b)`. Выведи ускорение.
4. **Проверь формы** на каждом шаге: `X.shape`, `X.mean(axis=...).shape` и форму результата. Напиши в markdown, почему broadcasting здесь сработал — с выравниванием справа налево.
5. **Убедись, что нормализация верна:** у нормализованных столбцов среднее должно быть ≈0, стандартное отклонение ≈1. Проверь и выведи.

1 Способ:

```python
def std_count(vect):
    total = 0
    for i in vect:
        total += (i - vect.mean())**2
    res = total / len(vect)
    res = res**(1/2)
    return res
X1 = np.zeros((1000, 5), dtype='float64')
start_time = time.perf_counter()
for j in range(5):
    mean = X.sum(axis=0)[j] / 1000
    std = std_count(X[:, j])
    for k in range(1000):
        X1[k, j] = (X[k, j] - mean) / std
stop_time = time.perf_counter()
time1 = stop_time - start_time
print(time1)
X1
```

2 Способ:

```python
start_time = time.perf_counter()
means = X.mean(axis=0)
stds = X.std(axis=0)
X2 = (X - means) / stds
stop_time = time.perf_counter()
time2 = stop_time - start_time
print(time2)
X2
```

Выбрал axis=0 потому что надо избавиться от 1000 в shape = (1000, 5) а 1000 на 0 индексе поэтому я убрал 0 ось, а broadcasting сработал ведь операции   происходили с (1000, 5) (5,) - совпадают 5 значит операции проводится

Проверяю совпадение и вывожу среднее и среднее отклонение:

```python
print(np.allclose(X1, X2))
print(f'Средние первого способа: {X1.mean(axis=0)}, {X1.std(axis=0)}')
print(f'Средние второго способа: {X1.mean(axis=0)}, {X1.std(axis=0)}')
```

По итогу вывожу во сколько раз нумпай быстрее:

```python
print(f'Numpy быстрее в {time1 / time2} раз')
```
